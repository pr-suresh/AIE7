#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.


##### ✅ Answer:

1. Memory caching :
    Pros: Extremely fast data access (nanoseconds to microseconds), suitable for active datasets and high-performance applications. Reduces latency and offloads the primary data source. Allows CPU direct access to data.
    Cons: Volatile (data lost on system restart), limited storage capacity compared to disk, more expensive to scale for large datasets. Risk of data loss during outages without persistence mechanisms.


    Disk caching:
    Pros: Persistent storage (data remains after restarts), larger capacity and generally cheaper than RAM, easier to implement and manage.
    Cons: Slower access times (milliseconds), limited scalability in terms of read/write speed under high traffic, increased wear and tear on the disk due to frequent access. 


2. Cache Invalidation strategies
    Determining when to invalidate or refresh cached data is crucial. Stale data can lead to incorrect responses, while excessive invalidation can degrade performance. Invalidating data in client-side caches is particularly difficult, as the server cannot directly control them. Choosing the right expiration and eviction policies is essential, balancing data freshness with avoiding unnecessary refreshes.

3. Concurrent access patterns
    Caches are excellent for reducing contention and latency when many users concurrently access the same data.
    Caches are not ideal for applications generating random or highly variable access patterns (e.g., sweeping through the keyspace of a large, frequently changing dataset). 

4. Cache size management
    Choosing the right cache size: Striking a balance between maximizing the cache hit rate and avoiding excessive memory consumption is critical. Too small a cache leads to a high miss rate and thrashing, while too large a cache can cause memory constraints and slowdowns.
    Eviction policies: Strategies like Least Recently Used (LRU) or Time-To-Live (TTL) are used to manage cache content and remove stale or less frequently accessed items. 

5. Cold start scenarios
    New server instances or cached data flushes can cause a burst of requests to the backend service.
    Strategies to mitigate this include request coalescing and cache warm-up techniques (preloading frequently accessed data during off-peak hours). 

When is caching most/least useful for production systems?

    Most useful
    Read-heavy workloads: Systems with significantly more read operations than writes benefit greatly from caching as it reduces the load on the primary data source.
    Frequently accessed, relatively static data: Caching is ideal for data that is accessed repeatedly but changes infrequently (e.g., product catalogs, popular articles).
    Latency-sensitive applications: Applications requiring low latency and high throughput (e.g., real-time analytics, online gaming) can significantly benefit from the speed of caching.
    Reducing network traffic: In distributed systems, caching local copies of remote data reduces network calls, improving performance, especially for geographically dispersed users.
    Scaling bottlenecks: Caching can alleviate pressure on databases or other backend services that struggle to handle high concurrency. 

    Least useful
    Write-heavy workloads: Systems with frequent updates to data struggle with cache consistency and the overhead of updating/invalidating caches can negate the performance benefits.
    Infrequently accessed or unique data: Caching data that is rarely accessed or unique per request provides little to no benefit and wastes cache resources.
    Small datasets: For small datasets, the time saved by caching may not justify the added complexity and overhead.
    Real-time systems requiring strict data consistency: When every piece of data needs to be absolutely up-to-date and consistent with the source, caching can introduce complexities and potential for stale data.


##### 🏗️ Activity #1: Cache Performance Testing

Create a simple experiment that tests our production caching system:

1. **Test embedding cache performance**: Try embedding the same text multiple times
2. **Test LLM cache performance**: Ask the same question multiple times  
3. **Measure cache hit rates**: Compare first call vs subsequent calls

````python
import time
import statistics
from typing import List, Dict, Any
import os

# Setup cache performance testing framework
class CachePerformanceTester:
    """Test cache performance for embeddings and LLM calls."""
    
    def __init__(self, rag_chain):
        self.rag_chain = rag_chain
        self.results = {}
    
    def time_function(self, func, *args, **kwargs) -> Dict[str, Any]:
        """Time a function call and return results with timing info."""
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        return {
            "result": result,
            "execution_time": end_time - start_time
        }
    
    def test_embedding_cache(self, test_text: str, num_runs: int = 5) -> Dict[str, Any]:
        """Test embedding cache performance by embedding the same text multiple times."""
        print(f"🧪 Testing Embedding Cache Performance")
        print(f"Text: '{test_text[:50]}...'")
        print(f"Number of runs: {num_runs}")
        print("-" * 50)
        
        times = []
        embeddings_instance = self.rag_chain.cached_embeddings.get_embeddings()
        
        for i in range(num_runs):
            result = self.time_function(embeddings_instance.embed_query, test_text)
            times.append(result["execution_time"])
            print(f"Run {i+1}: {result['execution_time']:.4f}s")
        
        # Calculate statistics
        first_call_time = times[0]
        subsequent_calls_avg = statistics.mean(times[1:]) if len(times) > 1 else 0
        cache_speedup = first_call_time / subsequent_calls_avg if subsequent_calls_avg > 0 else float('inf')
        
        results = {
            "test_type": "embedding_cache",
            "first_call_time": first_call_time,
            "subsequent_calls_avg": subsequent_calls_avg,
            "all_times": times,
            "cache_speedup": cache_speedup,
            "cache_hit_rate": (len(times) - 1) / len(times) * 100 if len(times) > 1 else 0
        }
        
        print(f"\n📊 Embedding Cache Results:")
        print(f"First call (cache miss): {first_call_time:.4f}s")
        if subsequent_calls_avg > 0:
            print(f"Subsequent calls avg (cache hits): {subsequent_calls_avg:.4f}s")
            print(f"Cache speedup: {cache_speedup:.2f}x faster")
        print(f"Cache hit rate: {results['cache_hit_rate']:.1f}%")
        
        return results
    
    def test_llm_cache(self, test_question: str, num_runs: int = 5) -> Dict[str, Any]:
        """Test LLM cache performance by asking the same question multiple times."""
        print(f"\n🤖 Testing LLM Cache Performance")
        print(f"Question: '{test_question[:50]}...'")
        print(f"Number of runs: {num_runs}")
        print("-" * 50)
        
        times = []
        responses = []
        
        for i in range(num_runs):
            result = self.time_function(self.rag_chain.invoke, test_question)
            times.append(result["execution_time"])
            responses.append(str(result["result"].content))
            print(f"Run {i+1}: {result['execution_time']:.4f}s")
        
        # Calculate statistics
        first_call_time = times[0]
        subsequent_calls_avg = statistics.mean(times[1:]) if len(times) > 1 else 0
        cache_speedup = first_call_time / subsequent_calls_avg if subsequent_calls_avg > 0 else float('inf')
        
        # Check if responses are identical (indicating cache hits)
        identical_responses = all(resp == responses[0] for resp in responses[1:])
        
        results = {
            "test_type": "llm_cache",
            "first_call_time": first_call_time,
            "subsequent_calls_avg": subsequent_calls_avg,
            "all_times": times,
            "cache_speedup": cache_speedup,
            "cache_hit_rate": (len(times) - 1) / len(times) * 100 if len(times) > 1 else 0,
            "identical_responses": identical_responses,
            "responses": responses
        }
        
        print(f"\n📊 LLM Cache Results:")
        print(f"First call (cache miss): {first_call_time:.4f}s")
        if subsequent_calls_avg > 0:
            print(f"Subsequent calls avg (cache hits): {subsequent_calls_avg:.4f}s")
            print(f"Cache speedup: {cache_speedup:.2f}x faster")
        print(f"Cache hit rate: {results['cache_hit_rate']:.1f}%")
        print(f"Identical responses: {identical_responses}")
        
        return results
    
    def run_comprehensive_test(self) -> Dict[str, Any]:
        """Run both embedding and LLM cache tests."""
        print("🚀 Starting Comprehensive Cache Performance Test")
        print("=" * 60)
        
        # Test data
        test_text = "What are the key requirements for federal student aid eligibility?"
        test_question = "What documents are needed to apply for federal student aid?"
        
        # Run tests
        embedding_results = self.test_embedding_cache(test_text)
        llm_results = self.test_llm_cache(test_question)
        
        # Combined analysis
        combined_results = {
            "embedding_test": embedding_results,
            "llm_test": llm_results,
            "overall_analysis": {
                "embedding_speedup": embedding_results["cache_speedup"],
                "llm_speedup": llm_results["cache_speedup"],
                "avg_cache_hit_rate": (embedding_results["cache_hit_rate"] + llm_results["cache_hit_rate"]) / 2
            }
        }
        
        print(f"\n🎯 Overall Cache Performance Summary:")
        print(f"Embedding cache speedup: {embedding_results['cache_speedup']:.2f}x")
        print(f"LLM cache speedup: {llm_results['cache_speedup']:.2f}x")
        print(f"Average cache hit rate: {combined_results['overall_analysis']['avg_cache_hit_rate']:.1f}%")
        
        return combined_results

# Initialize the tester with our RAG chain
cache_tester = CachePerformanceTester(rag_chain)

# Run the comprehensive cache performance test
test_results = cache_tester.run_comprehensive_test()
````

````python
##### 🏗️ Activity #2: Advanced Agent Testing

# Comprehensive LangGraph Agent Experimentation Framework
import time
from typing import List, Dict, Any
from langchain_core.messages import HumanMessage

class AgentExperimentFramework:
    """Framework for comprehensive agent testing and comparison."""
    
    def __init__(self, simple_agent):
        self.simple_agent = simple_agent
        self.results = []
        
    def time_agent_call(self, agent, query: str) -> Dict[str, Any]:
        """Time an agent call and return results with timing info."""
        start_time = time.time()
        try:
            messages = [HumanMessage(content=query)]
            response = agent.invoke({"messages": messages})
            end_time = time.time()
            
            # Extract final message and tool calls
            final_message = response["messages"][-1]
            tool_calls = []
            for msg in response["messages"]:
                if hasattr(msg, 'tool_calls') and msg.tool_calls:
                    tool_calls.extend([call['name'] for call in msg.tool_calls])
            
            return {
                "success": True,
                "response": final_message.content,
                "execution_time": end_time - start_time,
                "total_messages": len(response["messages"]),
                "tools_used": list(set(tool_calls)),
                "full_conversation": response["messages"]
            }
        except Exception as e:
            end_time = time.time()
            return {
                "success": False,
                "error": str(e),
                "execution_time": end_time - start_time,
                "response": None,
                "total_messages": 0,
                "tools_used": [],
                "full_conversation": []
            }
    
    def test_different_query_types(self):
        """Test different types of queries to observe tool selection patterns."""
        print("🧪 Testing Different Query Types")
        print("=" * 60)
        
        test_queries = {
            "factual_rag": {
                "query": "What are the eligibility requirements for federal student aid?",
                "expected_tools": ["retrieve_information"],
                "description": "Should favor RAG tool for document-based information"
            },
            "current_events": {
                "query": "What are the latest changes in student loan policies in 2024?",
                "expected_tools": ["tavily_search_results_json"],
                "description": "Should favor Tavily search for current information"
            },
            "academic_research": {
                "query": "Show me recent research papers about student debt impact on mental health",
                "expected_tools": ["arxiv"],
                "description": "Should favor Arxiv tool for academic research"
            },
            "multi_step": {
                "query": "What documents do I need for FAFSA and what recent studies show about financial aid effectiveness?",
                "expected_tools": ["retrieve_information", "arxiv"],
                "description": "Should use multiple tools for complex query"
            }
        }
        
        for query_type, details in test_queries.items():
            print(f"\n🔍 Testing: {query_type.upper()}")
            print(f"Query: {details['query']}")
            print(f"Expected tools: {details['expected_tools']}")
            print(f"Description: {details['description']}")
            print("-" * 50)
            
            result = self.time_agent_call(self.simple_agent, details['query'])
            
            if result["success"]:
                print(f"✅ Response generated successfully")
                print(f"⏱️ Time: {result['execution_time']:.2f}s")
                print(f"🛠️ Tools used: {result['tools_used']}")
                print(f"💬 Total messages: {result['total_messages']}")
                print(f"📝 Response preview: {result['response'][:150]}...")
                
                # Check if expected tools were used
                expected_used = any(tool in result['tools_used'] for tool in details['expected_tools'])
                print(f"🎯 Expected tool usage: {'✅' if expected_used else '❌'}")
            else:
                print(f"❌ Error: {result['error']}")
            
            # Store result for analysis
            self.results.append({
                "query_type": query_type,
                "query": details['query'],
                "result": result,
                "expected_tools": details['expected_tools']
            })
            
            print()
    
    def test_cache_performance_with_agents(self):
        """Test cache performance with repeated agent queries."""
        print("🚀 Testing Agent Cache Performance")
        print("=" * 50)
        
        test_query = "What documents are needed to apply for federal student aid?"
        num_runs = 3
        
        print(f"Query: {test_query}")
        print(f"Number of runs: {num_runs}")
        print("-" * 40)
        
        times = []
        for i in range(num_runs):
            result = self.time_agent_call(self.simple_agent, test_query)
            if result["success"]:
                times.append(result["execution_time"])
                print(f"Run {i+1}: {result['execution_time']:.2f}s | Tools: {result['tools_used']}")
            else:
                print(f"Run {i+1}: Failed - {result['error']}")
        
        if len(times) > 1:
            first_call = times[0]
            subsequent_avg = sum(times[1:]) / len(times[1:])
            speedup = first_call / subsequent_avg if subsequent_avg > 0 else float('inf')
            
            print(f"\n📊 Cache Performance Results:")
            print(f"First call (cache miss): {first_call:.2f}s")
            print(f"Subsequent calls avg: {subsequent_avg:.2f}s")
            print(f"Cache speedup: {speedup:.2f}x faster")
    
    def test_error_handling(self):
        """Test agent error handling and production readiness."""
        print("\n🛡️ Testing Error Handling & Production Readiness")
        print("=" * 60)
        
        error_scenarios = [
            {
                "name": "Invalid API query",
                "query": "Show me classified government documents about student loans",
                "description": "Should handle API restrictions gracefully"
            },
            {
                "name": "Extremely long query",
                "query": "What are the requirements for " + "student aid eligibility " * 100,
                "description": "Should handle token limits properly"
            },
            {
                "name": "Empty query",
                "query": "",
                "description": "Should handle empty input gracefully"
            }
        ]
        
        for scenario in error_scenarios:
            print(f"\n🔬 Testing: {scenario['name']}")
            print(f"Description: {scenario['description']}")
            
            result = self.time_agent_call(self.simple_agent, scenario['query'])
            
            if result["success"]:
                print(f"✅ Handled gracefully")
                print(f"⏱️ Time: {result['execution_time']:.2f}s")
                print(f"📝 Response preview: {result['response'][:100]}...")
            else:
                print(f"❌ Error occurred: {result['error']}")
    
    def analyze_results(self):
        """Analyze and summarize all test results."""
        print("\n📊 Comprehensive Analysis Results")
        print("=" * 60)
        
        if not self.results:
            print("No results to analyze")
            return
        
        # Tool usage analysis
        tool_usage = {}
        for result in self.results:
            for tool in result["result"].get("tools_used", []):
                tool_usage[tool] = tool_usage.get(tool, 0) + 1
        
        print(f"🛠️ Tool Usage Summary:")
        for tool, count in tool_usage.items():
            print(f"  - {tool}: {count} times")
        
        # Performance analysis
        successful_results = [r for r in self.results if r["result"]["success"]]
        if successful_results:
            avg_time = sum(r["result"]["execution_time"] for r in successful_results) / len(successful_results)
            avg_messages = sum(r["result"]["total_messages"] for r in successful_results) / len(successful_results)
            
            print(f"\n⚡ Performance Summary:")
            print(f"  - Average response time: {avg_time:.2f}s")
            print(f"  - Average conversation length: {avg_messages:.1f} messages")
            print(f"  - Success rate: {len(successful_results)}/{len(self.results)} ({len(successful_results)/len(self.results)*100:.1f}%)")
        
        # Tool selection accuracy
        accurate_selections = 0
        for result in self.results:
            if result["result"]["success"]:
                expected_tools = result["expected_tools"]
                used_tools = result["result"]["tools_used"]
                if any(tool in used_tools for tool in expected_tools):
                    accurate_selections += 1
        
        if successful_results:
            accuracy = (accurate_selections / len(successful_results)) * 100
            print(f"  - Tool selection accuracy: {accuracy:.1f}%")

# Initialize the experiment framework
if simple_agent:
    experiment_framework = AgentExperimentFramework(simple_agent)
    
    # Run comprehensive tests
    experiment_framework.test_different_query_types()
    experiment_framework.test_cache_performance_with_agents()
    experiment_framework.test_error_handling()
    experiment_framework.analyze_results()
else:
    print("⚠️ Simple agent not available - skipping comprehensive experiments")
````

````python
# Create a Helpfulness Agent with Response Evaluation

from typing import Optional, List, Dict, Any
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langchain_core.messages import BaseMessage, AIMessage, HumanMessage
from typing_extensions import TypedDict, Annotated
from langgraph.graph.message import add_messages

# Import from the agents module
from langgraph_agent_lib.agents import get_default_tools

class HelpfulnessAgentState(TypedDict):
    """Enhanced state schema for helpfulness agent."""
    messages: Annotated[List[BaseMessage], add_messages]
    helpfulness_score: float
    needs_refinement: bool
    refinement_count: int

def create_helpfulness_agent(
    model_name: str = "gpt-4.1-mini",
    temperature: float = 0.1,
    tools: Optional[List] = None,
    rag_chain: Optional[ProductionRAGChain] = None,
    max_refinements: int = 2
):
    """Create a helpfulness-checking LangGraph agent."""
    if tools is None:
        tools = get_default_tools(rag_chain)
    
    # Get model and bind tools
    model = get_openai_model(model_name=model_name, temperature=temperature)
    model_with_tools = model.bind_tools(tools)
    
    # Helpfulness evaluation prompt
    helpfulness_prompt = ChatPromptTemplate.from_messages([
        ("system", """You are a helpfulness evaluator. Rate the response on a scale of 0-10 based on:
        - Accuracy and factual correctness
        - Completeness of the answer
        - Relevance to the question
        - Clarity and usefulness
        
        Return only a number between 0-10. If the score is below 7, the response needs refinement."""),
        ("human", "Question: {question}\n\nResponse: {response}\n\nHelpfulness score:")
    ])
    
    helpfulness_model = get_openai_model(model_name="gpt-4.1-mini", temperature=0)
    helpfulness_chain = helpfulness_prompt | helpfulness_model | StrOutputParser()
    
    def call_model(state: HelpfulnessAgentState) -> Dict[str, Any]:
        """Invoke the model with messages."""
        messages = state["messages"]
        response = model_with_tools.invoke(messages)
        return {
            "messages": [response],
            "refinement_count": state.get("refinement_count", 0)
        }
    
    def evaluate_helpfulness(state: HelpfulnessAgentState) -> Dict[str, Any]:
        """Evaluate the helpfulness of the response."""
        messages = state["messages"]
        last_response = messages[-1]
        original_question = messages[0].content
        
        try:
            score_str = helpfulness_chain.invoke({
                "question": original_question,
                "response": last_response.content
            })
            score = float(score_str.strip())
        except:
            score = 5.0  # Default score if evaluation fails
        
        needs_refinement = (score < 7.0 and 
                          state.get("refinement_count", 0) < max_refinements)
        
        return {
            "helpfulness_score": score,
            "needs_refinement": needs_refinement
        }
    
    def refine_response(state: HelpfulnessAgentState) -> Dict[str, Any]:
        """Refine the response based on helpfulness evaluation."""
        messages = state["messages"]
        original_question = messages[0].content
        current_response = messages[-1].content
        current_score = state.get("helpfulness_score", 0)
        
        refinement_prompt = f"""The previous response scored {current_score}/10 for helpfulness. 
        Please provide a more helpful, accurate, and complete answer to: {original_question}
        
        Previous response: {current_response}
        
        Improved response:"""
        
        refined_message = HumanMessage(content=refinement_prompt)
        response = model_with_tools.invoke([refined_message])
        
        return {
            "messages": [response],
            "refinement_count": state.get("refinement_count", 0) + 1
        }
    
    def should_continue(state: HelpfulnessAgentState):
        """Route to tools if the last message has tool calls."""
        last_message = state["messages"][-1]
        if getattr(last_message, "tool_calls", None):
            return "action"
        return "evaluate"
    
    def should_refine(state: HelpfulnessAgentState):
        """Decide whether to refine the response."""
        if state.get("needs_refinement", False):
            return "refine"
        return END
    
    # Build enhanced graph
    graph = StateGraph(HelpfulnessAgentState)
    tool_node = ToolNode(tools)
    
    graph.add_node("agent", call_model)
    graph.add_node("action", tool_node)
    graph.add_node("evaluate", evaluate_helpfulness)
    graph.add_node("refine", refine_response)
    
    graph.set_entry_point("agent")
    graph.add_conditional_edges("agent", should_continue, {
        "action": "action", 
        "evaluate": "evaluate"
    })
    graph.add_edge("action", "agent")
    graph.add_conditional_edges("evaluate", should_refine, {
        "refine": "refine",
        END: END
    })
    graph.add_edge("refine", "evaluate")
    
    return graph.compile()

# Create the Helpfulness Agent
print("Creating Helpfulness Agent with Response Evaluation...")
try:
    helpfulness_agent = create_helpfulness_agent(
        model_name="gpt-4.1-mini",
        temperature=0.1,
        rag_chain=rag_chain,
        max_refinements=2
    )
    print("✓ Helpfulness Agent created successfully!")
    print("  - Model: gpt-4.1-mini")
    print("  - Features: Response evaluation, automatic refinement")
    print("  - Max refinements: 2")
    print("  - Tools: Tavily Search, Arxiv, RAG System")
except Exception as e:
    print(f"❌ Error creating helpfulness agent: {e}")
    helpfulness_agent = None
````

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses

````python
# 🛡️ Production-Safe LangGraph Agent with Guardrails Integration

import time
from typing import Dict, Any, List, Optional, Literal
from langchain_core.messages import BaseMessage, AIMessage, HumanMessage, SystemMessage
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from typing_extensions import TypedDict, Annotated
from langgraph.graph.message import add_messages

class GuardedAgentState(TypedDict):
    """Enhanced state schema for guardrails-protected agent."""
    messages: Annotated[List[BaseMessage], add_messages]
    input_validation_result: Dict[str, Any]
    output_validation_result: Dict[str, Any]
    guard_failures: List[str]
    refinement_count: int
    blocked_content: bool
    security_alerts: List[str]

class ProductionSafeAgent:
    """Production-safe LangGraph agent with integrated Guardrails validation."""
    
    def __init__(
        self,
        model_name: str = "gpt-4.1-mini",
        temperature: float = 0.1,
        tools: Optional[List] = None,
        rag_chain: Optional[Any] = None,
        max_refinements: int = 2,
        enable_guardrails: bool = True
    ):
        self.model_name = model_name
        self.temperature = temperature
        self.tools = tools or get_default_tools(rag_chain)
        self.max_refinements = max_refinements
        self.enable_guardrails = enable_guardrails and guardrails_available
        
        # Initialize model with tools
        self.model = get_openai_model(model_name=model_name, temperature=temperature)
        self.model_with_tools = self.model.bind_tools(self.tools)
        
        # Build the agent graph
        self.agent = self._build_agent_graph()
        
        print(f"✅ Production-Safe Agent initialized")
        print(f"   - Model: {model_name}")
        print(f"   - Guardrails: {'Enabled' if self.enable_guardrails else 'Disabled'}")
        print(f"   - Tools: {len(self.tools)} available")
        print(f"   - Max refinements: {max_refinements}")
    
    def _build_agent_graph(self) -> StateGraph:
        """Build the LangGraph with integrated guardrails."""
        
        def input_validation_node(state: GuardedAgentState) -> Dict[str, Any]:
            """Validate user input before processing."""
            if not self.enable_guardrails:
                return {"input_validation_result": {"passed": True, "reason": "Guardrails disabled"}}
            
            user_input = state["messages"][-1].content
            validation_result = self._validate_input(user_input)
            
            # Log security alerts
            security_alerts = state.get("security_alerts", [])
            if not validation_result["passed"]:
                security_alerts.append(f"Input blocked: {validation_result['reason']}")
            
            return {
                "input_validation_result": validation_result,
                "blocked_content": not validation_result["passed"],
                "security_alerts": security_alerts
            }
        
        def agent_reasoning_node(state: GuardedAgentState) -> Dict[str, Any]:
            """Main agent reasoning with tool calls."""
            if state.get("blocked_content", False):
                # Return blocked content message
                blocked_message = AIMessage(
                    content="I apologize, but I cannot process that request as it violates our content policies. Please rephrase your question to focus on student loan and financial aid topics."
                )
                return {"messages": [blocked_message]}
            
            # Normal agent processing
            messages = state["messages"]
            response = self.model_with_tools.invoke(messages)
            return {"messages": [response]}
        
        def output_validation_node(state: GuardedAgentState) -> Dict[str, Any]:
            """Validate agent output before returning to user."""
            if not self.enable_guardrails or state.get("blocked_content", False):
                return {"output_validation_result": {"passed": True, "reason": "Input already blocked or guardrails disabled"}}
            
            last_message = state["messages"][-1]
            if hasattr(last_message, 'content') and last_message.content:
                validation_result = self._validate_output(last_message.content)
                
                # Log security alerts for output issues
                security_alerts = state.get("security_alerts", [])
                if not validation_result["passed"]:
                    security_alerts.append(f"Output blocked: {validation_result['reason']}")
                
                return {
                    "output_validation_result": validation_result,
                    "security_alerts": security_alerts
                }
            
            return {"output_validation_result": {"passed": True, "reason": "No content to validate"}}
        
        def refinement_node(state: GuardedAgentState) -> Dict[str, Any]:
            """Refine output that failed validation."""
            refinement_count = state.get("refinement_count", 0)
            failed_output = state["messages"][-1].content
            failure_reason = state["output_validation_result"]["reason"]
            
            refinement_prompt = f"""The previous response failed validation due to: {failure_reason}
            
Please provide a refined response that:
1. Stays strictly on the topic of student loans and financial aid
2. Contains no potentially harmful content
3. Is factual and helpful
4. Does not include sensitive information

Original response: {failed_output}

Refined response:"""
            
            refined_message = HumanMessage(content=refinement_prompt)
            response = self.model.invoke([refined_message])
            
            return {
                "messages": [response],
                "refinement_count": refinement_count + 1
            }
        
        def should_validate_input(state: GuardedAgentState) -> Literal["validate_input", "agent"]:
            """Route to input validation or skip if guardrails disabled."""
            return "validate_input" if self.enable_guardrails else "agent"
        
        def should_continue_to_tools(state: GuardedAgentState) -> Literal["action", "validate_output"]:
            """Route to tools if the last message has tool calls."""
            if state.get("blocked_content", False):
                return "validate_output"
            
            last_message = state["messages"][-1]
            if getattr(last_message, "tool_calls", None):
                return "action"
            return "validate_output"
        
        def should_refine_output(state: GuardedAgentState) -> Literal["refine", END]:
            """Decide whether to refine failed output."""
            if not self.enable_guardrails:
                return END
            
            output_validation = state.get("output_validation_result", {})
            refinement_count = state.get("refinement_count", 0)
            
            if (not output_validation.get("passed", True) and 
                refinement_count < self.max_refinements):
                return "refine"
            return END
        
        # Build the graph
        graph = StateGraph(GuardedAgentState)
        tool_node = ToolNode(self.tools)
        
        # Add nodes
        graph.add_node("validate_input", input_validation_node)
        graph.add_node("agent", agent_reasoning_node)
        graph.add_node("action", tool_node)
        graph.add_node("validate_output", output_validation_node)
        graph.add_node("refine", refinement_node)
        
        # Set entry point
        graph.set_entry_point("validate_input")
        
        # Add edges
        graph.add_conditional_edges("validate_input", 
                                   lambda state: "agent",
                                   {"agent": "agent"})
        
        graph.add_conditional_edges("agent", should_continue_to_tools, {
            "action": "action",
            "validate_output": "validate_output"
        })
        
        graph.add_edge("action", "agent")
        
        graph.add_conditional_edges("validate_output", should_refine_output, {
            "refine": "refine",
            END: END
        })
        
        graph.add_edge("refine", "validate_output")
        
        return graph.compile()
    
    def _validate_input(self, user_input: str) -> Dict[str, Any]:
        """Validate user input using guardrails."""
        try:
            # Check if guardrails are available and configured
            if not guardrails_available:
                return {"passed": True, "reason": "Guardrails not available"}
            
            # 1. Jailbreak detection
            if 'jailbreak_guard' in globals():
                jailbreak_result = jailbreak_guard.validate(user_input)
                if jailbreak_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Potential jailbreak attempt detected",
                        "guard": "jailbreak_detection"
                    }
            
            # 2. Topic restriction
            if 'topic_guard' in globals():
                topic_result = topic_guard.validate(user_input)
                if topic_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Query is off-topic (must be about student loans/financial aid)",
                        "guard": "topic_restriction"
                    }
            
            # 3. PII detection
            if 'pii_guard' in globals():
                pii_result = pii_guard.validate(user_input)
                if pii_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Sensitive information detected in input",
                        "guard": "pii_detection"
                    }
            
            return {"passed": True, "reason": "All input validations passed"}
            
        except Exception as e:
            print(f"⚠️ Input validation error: {e}")
            # If validation fails, block as a safety measure
            return {"passed": False, "reason": f"Validation error - blocking for safety: {str(e)}"}
    
    def _validate_output(self, agent_output: str) -> Dict[str, Any]:
        """Validate agent output using guardrails."""
        try:
            # Check if guardrails are available and configured
            if not guardrails_available:
                return {"passed": True, "reason": "Guardrails not available"}
            
            # 1. Content moderation (using profanity guard as content guard)
            if 'content_guard' in globals():
                content_result = content_guard.validate(agent_output)
                if content_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Inappropriate content detected in response",
                        "guard": "content_moderation"
                    }
            
            # 2. PII protection in output
            if 'pii_guard' in globals():
                pii_result = pii_guard.validate(agent_output)
                if pii_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Sensitive information detected in response",
                        "guard": "pii_protection"
                    }
            
            # 3. Topic adherence
            if 'topic_guard' in globals():
                topic_result = topic_guard.validate(agent_output)
                if topic_result.validation_passed is False:
                    return {
                        "passed": False,
                        "reason": "Response went off-topic",
                        "guard": "topic_adherence"
                    }
            
            return {"passed": True, "reason": "All output validations passed"}
            
        except Exception as e:
            print(f"⚠️ Output validation error: {e}")
            # If validation fails, trigger refinement for safety
            return {"passed": False, "reason": f"Validation error - refinement needed: {str(e)}"}
    
    def invoke(self, query: str, include_security_report: bool = False) -> Dict[str, Any]:
        """Invoke the guardrails-protected agent."""
        start_time = time.time()
        
        initial_state = {
            "messages": [HumanMessage(content=query)],
            "input_validation_result": {},
            "output_validation_result": {},
            "guard_failures": [],
            "refinement_count": 0,
            "blocked_content": False,
            "security_alerts": []
        }
        
        try:
            result = self.agent.invoke(initial_state)
            end_time = time.time()
            
            # Extract final response
            final_message = result["messages"][-1]
            
            response_data = {
                "response": final_message.content if hasattr(final_message, 'content') else str(final_message),
                "execution_time": end_time - start_time,
                "blocked": result.get("blocked_content", False),
                "refinement_count": result.get("refinement_count", 0),
                "total_messages": len(result["messages"])
            }
            
            if include_security_report:
                response_data["security_report"] = {
                    "input_validation": result.get("input_validation_result", {}),
                    "output_validation": result.get("output_validation_result", {}),
                    "security_alerts": result.get("security_alerts", []),
                    "guardrails_enabled": self.enable_guardrails
                }
            
            return response_data
            
        except Exception as e:
            return {
                "response": f"Error processing request: {str(e)}",
                "execution_time": time.time() - start_time,
                "blocked": True,
                "error": str(e)
            }

# Create the Production-Safe Agent
if guardrails_available:
    print("🛡️ Creating Production-Safe LangGraph Agent...")
    
    # Verify guardrails are available in the global scope
    missing_guards = []
    required_guards = ['topic_guard', 'jailbreak_guard', 'pii_guard', 'content_guard']
    
    for guard_name in required_guards:
        if guard_name not in globals():
            missing_guards.append(guard_name)
    
    if missing_guards:
        print(f"⚠️ Missing guardrails: {missing_guards}")
        print("📝 Note: Some guardrails may not be available, but the agent will still provide basic protection")
    
    safe_agent = ProductionSafeAgent(
        model_name="gpt-4.1-mini",
        temperature=0.1,
        rag_chain=rag_chain,
        max_refinements=2,
        enable_guardrails=True
    )
    print("✅ Production-Safe Agent ready for testing!")
    
    # Quick validation test
    print("\n🔍 Testing guardrails integration...")
    validation_test = safe_agent._validate_input("Tell me about cryptocurrency investments")
    print(f"   Input validation test: {'✅ Working' if not validation_test['passed'] else '⚠️ May need adjustment'}")
    print(f"   Reason: {validation_test['reason']}")
    
else:
    print("⚠️ Guardrails not available - creating agent without safety features")
    safe_agent = ProductionSafeAgent(
        model_name="gpt-4.1-mini",
        temperature=0.1,
        rag_chain=rag_chain,
        enable_guardrails=False
    )
````
